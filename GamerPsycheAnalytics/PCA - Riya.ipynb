### Import Libraries and load the dataset

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from scipy.stats import rankdata
import numpy as np
gp = pd.read_csv("/content/GPCleanedDataset.csv")
```

### Map rank_group to low, mid and high

```python
# Mapping Low/Mid/High to numbers
mapping = {'Low': 1, 'Mid': 2, 'High': 3}

# Create a new numeric column
gp['rank_group_num'] = gp['rank_group'].map(mapping)

# Preview the result
gp[['rank_group', 'rank_group_num']].head(20)
```

### Scaled Avg and max rank score from 0 - 10

```python
# Columns to scale
rank_columns = ['avg_rank_score', 'max_rank_score']

# Function to scale to 0-10
def scale_0_10(series):
    min_val = series.min()
    max_val = series.max()
    return 10 * (series - min_val) / (max_val - min_val)

  # Scale avg_rank_score and max_rank_score to 0-10
for col in ['avg_rank_score', 'max_rank_score']:
    gp[col + '_scaled'] = 10 * (gp[col] - gp[col].min()) / (gp[col].max() - gp[col].min())
    gp[col + '_scaled'] = gp[col + '_scaled'].round().astype(int)  # Round to nearest integer


```

### PCA on GAD

```python
cols = [
    'GAD1','GAD2','GAD3','GAD4','GAD5','GAD6','GAD7'
]
X = gp[cols]

X_scaled = StandardScaler().fit_transform(X)

pca = PCA()
pca.fit(X_scaled)

loadings = pd.DataFrame(
    pca.components_.T,
    index=cols,
    columns=[f'PC{i+1}' for i in range(len(cols))]
)

explained_variance = pca.explained_variance_ratio_

loadings, explained_variance
```

PCA checks whether multiple variables are measuring the same underlying thing

PCA shows one main component (PC1) explains most of the variance (~55%)

All GAD1–GAD7 load strongly on this same component

This means the GAD items all measure anxiety

You do not need to analyze each GAD item separately

You can safely use one score (e.g., GAD_T) in further analysis

This reduces redundancy and simplifies models
